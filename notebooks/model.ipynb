# Crowd Anomaly Detection - UCSD Pedestrian dataset

Learn what *normal* pedestrian motion looks like, then score how unusual new frames are.

**Pipeline:** frames -> dense optical flow (Farneback) -> per-cell motion statistics on a grid -> Gaussian-mixture model of normal motion -> anomaly score (negative log-likelihood) -> threshold.

The dataset is downloaded directly from Kaggle: [UCSD Anomaly Dataset](https://www.kaggle.com/datasets/orvile/ucsd-anomaly-dataset/data).
The reusable code lives in the `crowd_anomaly/` package; this notebook is the interactive walk-through.

## 1. Setup and dataset download

```python
import logging, sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd()))  # run from the repo root
import kagglehub
import matplotlib.pyplot as plt
import numpy as np

from crowd_anomaly import Config, CrowdAnomalyModel
from crowd_anomaly.data import (discover_clips, find_gt_m_file, find_subset_dir, iter_frames,
                                parse_gt_m_file)
from crowd_anomaly.evaluate import evaluate_clips, save_report
from crowd_anomaly.features import extract_many
from crowd_anomaly.inference import predict_source

logging.basicConfig(level=logging.INFO, format="%(levelname)s %(message)s")

# Download latest version
path = kagglehub.dataset_download("orvile/ucsd-anomaly-dataset")

print("Path to dataset files:", path)

SUBSET = "ped1"                           # "ped1" or "ped2"
MODEL_PATH = Path("models") / f"{SUBSET}.joblib"
N_JOBS = 1                                # -1 = use all CPU cores

subset_dir = find_subset_dir(path, SUBSET)
train_clips = discover_clips(subset_dir, "Train")
test_clips = discover_clips(subset_dir, "Test")
print(f"{len(train_clips)} training clips, {len(test_clips)} test clips in {subset_dir}")
```

## 2. Look at the data

```python
frames = list(iter_frames(test_clips[0]))
idx = [0, len(frames) // 2, len(frames) - 1]
fig, axes = plt.subplots(1, 3, figsize=(12, 3.2))
for ax, i in zip(axes, idx):
    ax.imshow(frames[i], cmap="gray"); ax.set_title(f"{test_clips[0].name} frame {i}"); ax.axis("off")
plt.tight_layout(); plt.show()
```

## 3. Configuration
All tunable settings live in one `Config` (it is stored inside the saved model).

```python
cfg = Config(
    grid_rows=6, grid_cols=8,   # spatial pooling grid
    n_components=8,             # GMM components (or set auto_components=True to pick by BIC)
    top_k=3,                    # frame score = mean of the k most unusual cells
    smooth_sigma=2.0,           # temporal smoothing of the frame score
    val_fraction=0.2,           # training clips held out to calibrate the threshold
    threshold_percentile=99.0,
    seed=0,
)
cfg
```

## 4. Extract optical-flow features
One `(rows, cols, 3)` array of `[mean dx, mean dy, mean magnitude]` per frame pair - computed **per clip**, so flow is never taken across clip boundaries.

```python
train_feats, frame_size = extract_many(train_clips, cfg, n_jobs=N_JOBS)
print("frame size:", frame_size, "| per-clip feature shape:", train_feats[0].shape)
```

```python
# Average motion energy on the grid: shows where people normally walk.
mean_mag = np.mean([f[..., 2].mean(axis=0) for f in train_feats], axis=0)
plt.imshow(mean_mag, cmap="magma"); plt.colorbar(label="mean flow magnitude")
plt.title("Where normal motion happens (training average)"); plt.show()
```

## 5. Train the normality model
The model sees **only normal clips**. Whole clips (never single frames) are held out to calibrate the threshold.

```python
model = CrowdAnomalyModel(cfg).fit(train_feats, frame_size)
print("components:", model.n_components_, "| frame threshold:", round(model.threshold_, 3))
model.save(MODEL_PATH)
print("saved ->", MODEL_PATH)
```

## 6. Evaluate on the test clips
Frame-level ROC-AUC / EER (pooled over clips), plus precision/recall at the calibrated threshold.

```python
gt_file = find_gt_m_file(subset_dir)
gt = parse_gt_m_file(gt_file) if gt_file else None
result = evaluate_clips(model, test_clips, gt, n_jobs=N_JOBS)
metrics = {k: v for k, v in result.items() if k not in ("clips", "threshold")}
for k, v in metrics.items():
    print(f"{k:>22}: {v:.4f}" if isinstance(v, float) else f"{k:>22}: {v}")
```

```python
fig, axes = plt.subplots(min(len(result["clips"]), 4), 1, figsize=(9, 7), squeeze=False)
for ax, c in zip(axes[:, 0], result["clips"]):
    ax.plot(c["scores"]); ax.axhline(model.threshold_, color="r", ls="--")
    ax.fill_between(np.arange(len(c["labels"])), 0, 1, where=c["labels"].astype(bool),
                    transform=ax.get_xaxis_transform(), color="orange", alpha=0.3)
    ax.set_ylabel(c["clip"], rotation=0, ha="right")
plt.suptitle("Anomaly score vs ground truth (orange)"); plt.tight_layout(); plt.show()
save_report(result, "reports")  # writes metrics.json, roc.png, timelines.png
```

## 7. Predict on a clip or video
`predict_source` accepts a video file (`.avi`, `.mp4`, ...) or a folder of frames, and writes `scores.csv`, `scores.png` and an annotated video.

```python
out = predict_source(model, test_clips[0], "outputs/example")
print("anomalous segments (frame ranges):", out["segments"])
print("wrote:", {k: str(out[k]) for k in ("csv", "plot", "video") if k in out})
```
